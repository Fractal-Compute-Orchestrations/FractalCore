```python
import numpy as np
from fractal_quant import quantize_tensor_int4, dequantize_tensor_int4, cosine_similarity

rng = np.random.default_rng(42)
weights = rng.normal(0.0, 0.02, size=(1024, 256)).astype(np.float32)
weights[0, :4] = 8.5

packet, stats = quantize_tensor_int4(weights, block_size=64)
recovered = dequantize_tensor_int4(packet)
print(f'Compression: {stats.compression_ratio:.2f}x | FRQ4 Cosine: {cosine_similarity(weights, recovered):.4f}')
```
